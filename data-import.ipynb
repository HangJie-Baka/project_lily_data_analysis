```python
from pymongo import MongoClient

client = MongoClient('mongodb://localhost:27017/project-lily')
```

```python
print(client.list_database_names())
```

```python
db = client['project-lily']
print(db.list_collection_names())
```

```python
chats = db['chats']
print(chats.count_documents({}))
```

```python
import json
chat_data = list(chats.find({}))
with open('chat_history.json', 'w', encoding='utf-8') as f:
    json.dump(chat_data, f, ensure_ascii=False, indent=4, default=str)

print(f"Exported {len(chat_data)} chat documents.")
```

```python
import pandas as pd
```

```python
df = pd.json_normalize(chat_data)
```

```python
df.sample(10, replace=True)
```

```python
df.to_csv('chat_data.csv', mode='w')
```

```python
user_message = []

for chat in chat_data:
    for message in chat['messages']:
        if message['role'] == "user":
            # print(message['content'])
            user_message.append(message['content'])

print(user_message)
```

```python
user_df = pd.DataFrame(user_message, columns=['message'])
user_df.to_csv('user_messages.csv', mode='w')
```
